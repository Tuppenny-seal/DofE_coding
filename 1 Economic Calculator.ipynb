```python
def percentage_change(old_value, new_value):
    percentage_change = ((new_value - old_value) / old_value)*100
    return percentage_change
```

```python
# growth rate as percentage
def compound_growth(initial_value, growth_rate, number_of_periods):
    final_value = initial_value * ((1+(growth_rate/100))**number_of_periods)
    return final_value
```

```python
# decay rate as percentage
def compound_decay(initial_value, decay_rate, number_of_periods):
    final_value = initial_value * ((1-(decay_rate/100))**number_of_periods)
    return final_value
```

```python
def gdp_per_person(GDP, population):
    gdp_per_person = GDP/population
    return gdp_per_person
```

```python
# inflation rate as percentage
def real_value(nominal_value, infaltion_rate, years):
    real_value = nominal_value /((1 + (infaltion_rate/100))**years)
    return real_value
```

```python
# inflation rate as percentage
def inflation_adjusted_value(original_value, infaltion_rate, years):
    adjusted_value = original_value *((1 + (infaltion_rate/100))**years)
    return adjusted_value
```

```python
def weekly_wage(hourly_wage, hours_worked):
    weekly_wage = hourly_wage*hours_worked
    return weekly_wage
```

```python
def annual_wage(weekly_wage):
    annual_wage = weekly_wage * 52
    return annual_wage
```
