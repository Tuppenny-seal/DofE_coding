# Read .csv from Multiple URLs with Pandas

Link: root + season + league

```python
# link structure
"https://football-data.co.uk/mmz4281/" + "2627" + "/" + "E0" + ".csv"
```

```python
# create a root variable
root = "https://football-data.co.uk/mmz4281/"
```

## Multiple leagues

https://football-data.co.uk/mmz4281/2627/E0.csv
https://football-data.co.uk/mmz4281/2627/E1.csv
https://football-data.co.uk/mmz4281/2627/E2.csv
https://football-data.co.uk/mmz4281/2627/E3.csv
https://football-data.co.uk/mmz4281/2627/EC.csv

```python
# creating list of leagues
leagues = ['E0', 'E1', 'E2', 'E3', 'EC']
frames = []

# looping through leagues, read multiple CSVs and append it into a list
for league in leagues:
    df = pd.read_csv(root + '2627' + '/' + league + '.csv')
    frames.append(df)
```

```python
# length of frames
len(frames)
```

```python
# show 2nd element
frames[1]
```

```python
#show last element
frames[-1]
```

## Multiple Seasons

```python
for season in range(20, 26):
    print(str(season) + str(season+1))
```

```python
leagues = ['E0', 'E1', 'E2', 'E3', 'EC']
frames = []

for league in leagues:
    for season in range(20, 26):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + league + '.csv', encoding='unicode_escape')
        df.insert(1, 'season', str(season) + str(season+1))
        frames.append(df)
```

```python
# total frames (6 seasons x 5 leagues)
len(frames)
```

```python
# show first element
frames[0]
```

```python
# show last element
frames[29]
```

# Organise all the data in a dictionary

```python
# create a dictionary with original name of league as key
dict_countries = {'Spanish La Liga':'SP1', 'Spanish Segunda Division':'SP2', 'German Bundesliga':'D1', 'Italian Serie A':'I1', 'English Premier League':'E0', 'English Championship':'E1', 'English League 1':'E2', 'French Ligue 1':'F1'}
```

```python
dict_countries['Spanish Segunda Division']
```

```python
for league in dict_countries:
    print(dict_countries[league])
```

```python
dict_historical_data = {}

for league in dict_countries:
    frames = []
    for season in range(20, 26):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + dict_countries[league] + '.csv', encoding='unicode_escape')
        df.insert(1, 'season', str(season) + str(season+1))
        frames.append(df)
    df_concat = pd.concat(frames)
    dict_historical_data[league] = df_concat
```
