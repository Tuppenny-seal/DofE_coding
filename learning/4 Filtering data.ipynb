# Creating a Conditional Column for more than 2 choices with np.select()

```python
# create an array based on multiple price tiers (2+ choices)

# create "conditions" and "values" variables
conditions = [df_laptops['Price_euros']>3000,
              (df_laptops['Price_euros']<=3000) & (df_laptops['Price_euros']>2000),
              (df_laptops['Price_euros']>1000) & (df_laptops['Price_euros']<= 2000), 
              df_laptops['Price_euros']<=1000]
values = ['Very Expensive', 'Expensive', 'Affordable', 'Cheap']
```

```python
# set it to a new column
df_laptops['Price_tier'] = np.select(conditions, values, default="Unknown")
```

```python
df_laptops
```

```python
df_laptops['Price_tier'].value_counts()
```

```python
df_laptops[~duplicated_first][['Company', 'Price_euros']]
```

```python
df_laptops[~duplicated_first].value_counts('Company')
```

```python
# keep=last (most expensive laptops per company
duplicated_last = df_laptops.duplicated('Company', keep='last')
```

```python
df_laptops[~duplicated_last]
```

```python
df_laptops[~duplicated_last][['Company', 'Price_euros']]
```

```python
# keep=False
duplicated_False = df_laptops.duplicated('Company', keep=False)
```

```python
df_laptops[~duplicated_False]
```

```python
# empty dataframe as all elements in company column are repeated
df_laptops.value_counts('Company')
```

# Drop Duplicate Elements with the drop_duplicates() method

```python
# drop duplicates in multiple columns
df_laptops.drop_duplicates(['Company'])
```

```python
df_laptops.drop_duplicates(['Company']).value_counts("Company")
```

```python
# sort dataframe asending by company and price (cheapest first, most expensive last)
df_laptops1 = df_laptops.sort_values(['Company', 'Price_euros']).copy()
```

```python
# cheapest: keep='first'
df_laptops1.drop_duplicates(['Company'], keep='first')[['Company', 'Price_euros']]
```

```python
# most expernsive (keep='last')
df_laptops1.drop_duplicates(['Company'], keep='last')[['Company', 'Price_euros']]
```

```python
# Arguements 
# ignore_index=True: the resulting axis will be labelled 0, 1, ...
df_laptops1.drop_duplicates(['Company'], keep='last', inplace=True, ignore_index=True)
```

```python
df_laptops1[['Company', 'Price_euros']]
```
