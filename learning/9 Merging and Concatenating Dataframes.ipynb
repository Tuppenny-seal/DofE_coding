```python
import pandas as pd
```

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)[['imdb_title_id', 'title', 'year', 'genre', 'country']]
df_ratings = pd.read_csv('IMDb ratings.csv')[['imdb_title_id', 'total_votes', 'mean_vote']]
```

# Concat()

## Concatenate vertically

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['E', 'F', 'G', 'F'],
                    'age': [40, 21, 19, 24]})
```

```python
print(df1)
print(df2)
```

```python
df = pd.concat([df1, df2], axis=0, ignore_index=True)
```

```python
df
```

# merge()

```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'],
                    'age': [30, 23, 25, 22]})
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'],
                    'job': ['Doctor', 'Statistician', 'Accountant', 'Developer']})
```

```python
print(df1)
print(df2)
```

## Inner join

```python
df_inner = df1.merge(df2, on='id', how='inner')
```

```python
df_inner
```

```python
#shapes
print(df1.shape)
print(df2.shape)
print(df_inner.shape)
```

## Exclusive Outer Join

```python
df1
```

```python
df2
```

```python
# indicator=True makes '_merge' column
df1.merge(df2, on='id', how='outer', indicator=True)
```

```python
# query
df_exclusive = df1.merge(df2, on='id', how='outer', indicator=True
                        ).query("_merge=='left_only' | _merge=='right_only'")
```

```python
df_exclusive
```

```python
df_exclusive.shape
```
