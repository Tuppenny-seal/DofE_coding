## Adding a new column with assign()

```python
# creating random numbers for new score columns
score1 = np.random.randint(0, 100, size=1000)
score2 = np.random.randint(0, 100, size=1000)
```

```python
# creating a series using the random numbers
series1 = pd.Series(score1, index=np.arange(0, 1000))
series2 = pd.Series(score2, index=np.arange(0, 1000))
```

```python
series1
```

```python
# using assign() to add multiple columns
df_exams = df_exams.assign(score1=series1, score2=series2)
```

```python
df_exams
```

# Operations on Dataframes

## Math Operations

### Operations in columns

```python
# Selecting a column and calculating total sum
df_exams['math score'].sum()
```

```python
# count, mean, std, max, min
print(df_exams['math score'].count())
print(df_exams['math score'].mean())
print(df_exams['math score'].std())
print(df_exams['math score'].max())
print(df_exams['math score'].min())
```

```python
# easier calculation with .decsribe
df_exams['math score'].describe()
```

### Operating in rows

```python
# calculating the sum in a row
df_exams['math score'] + df_exams['reading score'] + df_exams['writing score']
```

```python
# calculating the average score and assgning the result to a new column
df_exams['average score'] = (df_exams['math score'] + df_exams['reading score'] + df_exams['writing score'])/3
```

```python
df_exams = df_exams.round(2)
df_exams
```

# The value_counts() method

```python
# len function
len(df_exams['gender'])
```

```python
# .count() method
df_exams['gender'].count()
```

```python
# counting gender elements
df_exams['gender'].value_counts()
```

```python
# return the relative frequency (divide all values by the sum of all values)
df_exams['gender'].value_counts(normalize=True)
```

```python
# counting "parental level of education" elements by catagory
df_exams['parental level of education'].value_counts()
```

```python
# return relative frequency and round to 2 dp
df_exams['parental level of education'].value_counts(normalize=True).round(2)
```

# Sort a dataframe with sort_values()

```python
# sort by one column
df_exams.sort_values(by='math score')
```

```python
# sort descending by one column
df_exams.sort_values('math score', ascending=False)
```

```python
df_exams
```

```python
# sort by descending by multiple columns 
df_exams.sort_values(['math score', 'reading score'], ascending=False, inplace=True)
```

```python
df_exams
```

```python
# sort descending with key function
df_exams.sort_values('race/ethnicity', ascending=True, key=lambda col:col.str.lower())
```

# set_index() and sort_index() methods

## Create Index

```python
import random
```

```python
# creating non-repetitive values for the index
new_index = np.arange(0, 1000)
```

```python
# shuffling the index 
random.shuffle(new_index)
```

```python
# create a new column with new index
df_exams['new_index'] = new_index
df_exams
```

## set index

```python
# set new_index column as index
df_exams.set_index('new_index', inplace=True)
df_exams
```

## sort index

```python
# sort dataframe by new index
df_exams.sort_index(inplace=True)
df_exams
```

```python
# sort descending my index
df_exams.sort_index(ascending=False)
```
