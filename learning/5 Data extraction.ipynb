# Selecting elements by index lable with loc()

## Selecting with a single value

loc[row_label, column_label]

```python
# get all data about L.Messi
df.loc['L. Messi']
```

```python
# get the height of L.Messi
df.loc['L. Messi', 'height_cm']
```

```python
# get the weight of Cristiano Ronaldo
df.loc['Cristiano Ronaldo', 'weight_kg']
```

```python
# get all rows inside the 'height_cm' column
df.loc[:,'height_cm']
```

```python
# get all the columns that correspond to the index 'L. Messi'
df.loc['L. Messi', :]
```

## Selecting with a list of values

```python
# get all data about L. Messi and Cristiano Ronaldo
df.loc[['L. Messi', 'Cristiano Ronaldo'], :]
```

```python
# get the height of L. Messi and Cristiano Ronaldo
df.loc[['L. Messi', 'Cristiano Ronaldo'], 'height_cm']
```

```python
# get the height and weight of L. Messi
df.loc['L. Messi', ['height_cm', 'weight_kg']]
```

```python
# get height and weight of L. Messi and Cristiano Ronaldo
df.loc[['L. Messi', 'Cristiano Ronaldo'], ['height_cm', 'weight_kg']]
```

## Selecting range of data with a slice

start:stop:step (Note: both start and stop are included, contrary to usual python slices)

```python
# slice column labels
players = ['L. Messi', 'Cristiano Ronaldo']
df.loc[players, 'age':'club']
```

```python
# slice index labels
columns = ['age', 'dob', 'height_cm', 'weight_kg']
```

```python
# get top 1 and top 10 player name
df.index[:10]
```

```python
df.loc['L. Messi':'M. Salah', columns]
```

## Selecting with conditions

```python
# one condition: select players with height above 180cm
df['height_cm']>180
```

```python
df.loc[df['height_cm']>180, columns]
```

```python
# multiple conditions: select players with height above 180cm and from Argentina
(df['height_cm']>180) & (df['nationality']=='Argentina')
```

```python
df.loc[(df['height_cm']>180) & (df['nationality']=='Argentina'), :]
```

# Selecting elements by index position with .iloc()

```python
df = pd.read_csv('players_20.csv')
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df
```

## Selecting with single value

iloc[row_position, column_position]

```python
# get the height of L. Messi
df.iloc[0, 3]
```

```python
# get the weight of Cristiano Ronaldo
df.iloc[1, 4]
```

```python
# get all rows inside the height_cm column
df.iloc[:, 3]
```

```python
# get all columns that correspond to the index L. Messi
df.iloc[0]
```

## Selecting with a list of values

```python
# get all data about L. Messi and Cristiano Ronaldo
df.iloc[[0,1]]
```

```python
# get the height of L. Messi and Cristiano Ronaldo
df.iloc[[0,1], 3]
```

```python
# get the height and weight of L. Messi
df.iloc[0, [3,4]]
```

```python
# get the height and weight of L. Messi and Cristiano Ronaldo
df.iloc[[0,1], [3,4]]
```

## Selecting a range of data with a slice

start:stop (note: start is included but stop is excluded)

```python
# slice column labels age/club
players = [0, 1]
df.iloc[players, 1:6+1]
```

```python
# slice index labels (top1 to top10 players)
columns = [1,2,3,4]
df.iloc[0:10, columns]
```

## Selecting with conditions

iloc cannot accept a boolean Series, but only a boolean list. We have to use the list() function to convert a series to a boolean list

```python
# one condition: select a player with height above 180cm
list(df['height_cm']>180)
```

```python
df.iloc[list(df['height_cm']>180), columns]
```

```python
# multiple conditions: select players with height above 180cm from Argentina
df.iloc[list((df['height_cm']>180)&(df['nationality']=='Argentina')), columns]
```

# Creating Random Sample with sample() Method

## sample()

```python
# extract 10 random elements from the "nationality" column
df['nationality'].sample(10, random_state=99)
```

```python
# extract a random 20% sample of the dataframe
df.sample(frac=0.2, random_state=52)
```

# Filter a Dataframe with the query method

```python
df = pd.read_csv('players_20.csv')
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# select players older than 32
df.query("age>34")
```

```python
# exercise: equivalent boolean slicing
df[df['age']>34]
```

```python
# select players older than 34 from Italy
df.query("age>34 & nationality=='Italy'")
```

```python
# exercise: equivalent boolean slicing
df[(df['age']>34)&(df['nationality']=='Italy')]
```

```python
# add a not operator to first example
df.query("not(age>34)")
```

```python
# exercise: equivalent boolean slicing
df[~(df['age']>34)]
```

```python
# convert height to meters and select those with height above 1.8
df.query("height_cm/100 > 1.8")
```

```python
# exercise: equivalent boolean slicing
df[df['height_cm']/100 > 1.8]
```

# The apply() method

```python
df = pd.read_csv('players_20.csv')
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
df['age']
```

```python
# use numpy function (square root) and apply it to a series
df['age'].apply(np.sqrt)
```

```python
# create your own function and apply it to a dataframe
def calculate_bmi(row):
    return row['weight_kg']/(row['height_cm']/100)**2
```

```python
# axis=1 means we can perform operations with multiple columns
df.apply(calculate_bmi, axis=1)
```

# Lambda function + apply() method

```python
df = pd.read_csv('players_20.csv')
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

## Lambda function

```python
# basic function
def sum_values(a, b):
    x = a+b
    return x
```

```python
sum_values
```

```python
sum_values(2,3)
```

```python
# lambda function (one liner)
sum_values_lambda = lambda a,b:a+b
```

```python
sum_values_lambda(2,3)
```

## Apply + Lambda function

```python
# use lambda function to convert "height_cm" to metres
df['height_cm'].apply(lambda x:x/100)
```

```python
# alternative
df['height_cm']/100
```

```python
# use lambda function to convert "long_name" serries to uppercase
df["long_name"].apply(lambda x:x.upper())
```

```python
# alternative with str attribute
df["long_name"].str.upper()
```

```python
df.dtypes
```

```python
df.dtypes
```

```python
# use lambda function to get year of "dob" series
df['dob'].apply(lambda x:x.year)
```

```python
# alternative with dt attribute
df['dob'].dt.year
```

```python
# apply lambda function to dataframe to calculate bmi
df.apply(lambda x:x['weight_kg']/(x['height_cm']/100)**2, axis=1)
```

# Make a copy of a Dataframe with copy()

```python
df = pd.read_csv('players_20.csv')
df.set_index('short_name', inplace=True)
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

## deep = True

```python
# df=True by default (modifications to copy don'effect original dataframe)
df_copy = df.copy()
```

```python
# update value in orginal dataframe
df.loc['L. Messi', 'height_cm'] = 180
```

```python
# copy vs original dataframe (different)
df
```

```python
df_copy
```

## deep = False

```python
# deep=False (any changes to the data of original will be reflected in the shallow copy and vise versa)
df_shallow_copy = df.copy(deep=False)
```

```python
# update value in original dataframe
df.loc['Cristiano Ronaldo', 'height_cm'] = 200
```

```python
# shallow copy vs original dataframe (same)
df
```

```python
df_shallow_copy
```
