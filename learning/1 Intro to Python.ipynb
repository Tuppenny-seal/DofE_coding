# For loop

```python
countries
```

```python
for country in countries:
    print(country)
```

```python
for country in enumerate(countries):
    print(country)
```

```python
for i, country in enumerate(countries):
    print(i)
    print(country)
```

```python
my_data = {'name':'Oscar', 'age':15}
my_data
```

```python
for key, value in my_data.items():
    print(key)
    print(value)
```
