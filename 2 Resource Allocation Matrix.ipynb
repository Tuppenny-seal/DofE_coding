```python
import numpy as np
```

```python
import pandas as pd
```

```python
resources = pd.DataFrame([[2, 6], [1, 4]], index=['Labour', 'Lithium'], columns=['Food', 'Electric Car'])
```

```python
resources
```

```python
resources['Electric Car']
```

```python
Availability = {'Labour':120, 'Lithium':80}
```

```python
plan_a = {'Food':40, 'Electric Cars':6}
```

```python
plan_a_resources = resources*plan_a.values()
```

```python
plan_a_resources['Total'] = plan_a_resources['Food'] + plan_a_resources['Electric Car']
```

```python
plan_a_resources
```

```python
plan_b = {'Food':18, 'Electric Cars':14}
```

```python
plan_b_resources = resources*plan_b.values()
```

```python
plan_b_resources['Total'] = plan_b_resources['Food'] + plan_b_resources['Electric Car']
```

```python
plan_b_resources
```

```python
plan_c = {'Food':30, 'Electric Cars':11}
```

```python
plan_c_resources = resources*plan_c.values()
```

```python
plan_c_resources['Total'] = plan_c_resources['Food'] + plan_c_resources['Electric Car']
```

```python
plan_c_resources
```

```python
plans_resources = pd.DataFrame(data=[plan_a_resources['Total'], plan_b_resources['Total'], plan_c_resources['Total']],
                                columns=['Labour', 'Lithium'], index=['Plan A', 'Plan B', 'Plan C'])
```

```python
plans_resources
```

```python
plans_resources['Labour']>120
```

```python
plans_resources['Lithium']>80
```

```python
(plans_resources['Labour']>120) | (plans_resources['Lithium']>80)
```

```python
np.where((plans_resources['Labour']>Availability['Labour']) | (plans_resources['Lithium']>Availability['Lithium']), 'Unfeasable', 'Feasable')
```

```python
plans_resources['Feasability'] = np.where((plans_resources['Labour']>120) | (plans_resources['Lithium']>80), 'Unfeasable', 'Feasable')
```

```python
plans_resources
```
